```python
import spacy

# en_core_web_trf - same parser as mean-mdl.ipynb, so all tree features come from the same parse
# only pos_ (tagger + attribute_ruler) and dep_/head/sents (parser) are needed, so ner and lemmatizer are disabled
spacy.prefer_gpu() # uses the Mac GPU (MPS) if available
nlp = spacy.load("en_core_web_trf", disable=["ner", "lemmatizer"])
print(nlp.meta["lang"] + "_" + nlp.meta["name"], nlp.meta["version"], nlp.pipe_names) # record in run notes

ARG_DEPS = {
    "nsubj",
    "nsubj:pass",
    "obj",
    "iobj",
    "csubj",
    "csubj:pass",
    "ccomp",
    "xcomp",
    "expl",
}
ADJUNCT_DEPS = {
    "obl",
    "obl:agent",
    "obl:tmod",
    "obl:npmod",
    "advmod",
    "advcl",
    "nmod",
    "nmod:poss",
    "nmod:tmod",
    "acl",
    "acl:relcl",
}
DEP_REL_TYPES = [
    "acl",
    "acl:relcl",
    "advcl",
    "advmod",
    "amod",
    "appos",
    "aux",
    "aux:pass",
    "case",
    "cc",
    "ccomp",
    "compound",
    "conj",
    "cop",
    "csubj",
    "dep",
    "det",
    "discourse",
    "expl",
    "fixed",
    "flat",
    "iobj",
    "mark",
    "nmod",
    "nmod:poss",
    "nsubj",
    "nsubj:pass",
    "nummod",
    "obj",
    "obl",
    "obl:npmod",
    "parataxis",
    "punct",
    "xcomp",
]
DEP_BIGRAM_KEYS = [
    "VERB_ADV_before",
    "VERB_ADV_after",
    "VERB_PRON_before",
    "VERB_PRON_after",
    "VERB_ADP_before",
    "VERB_ADP_after",
    "VERB_PROPN_before",
    "VERB_PROPN_after",
    "VERB_SCONJ_before",
    "VERB_SCONJ_after",
    "VERB_NUM_before",
    "VERB_NUM_after",
    "PRON_NOUN_before",
    "PRON_NOUN_after",
    "PRON_VERB_before",
    "PRON_VERB_after",
    "PROPN_PROPN_before",
    "PROPN_PROPN_after",
    "PROPN_NUM_before",
    "PROPN_NUM_after",
    "ADJ_SCONJ_before",
    "ADJ_SCONJ_after",
    "NOUN_DET_before",
    "NOUN_DET_after",
    "NOUN_ADJ_before",
    "NOUN_ADJ_after",
]

feature_names = ["syn_depth", "syn_isc", "arg_adjunct_ratio"]
for dep in DEP_REL_TYPES:
    feature_names.append("dep_" + dep.replace(":", "_"))
for key in DEP_BIGRAM_KEYS:
    feature_names.append("depbig_" + key)

def tree_depth(token):
    max_depth = 1
    stack = [(token, 1)]
    seen = set()
    while stack:
        node, depth = stack.pop()
        if depth > max_depth:
            max_depth = depth
        if depth >= 256:
            continue
        for child in node.children:
            edge = (node.i, child.i)
            if edge in seen:
                continue
            if child.i == node.i:
                continue
            seen.add(edge)
            stack.append((child, depth + 1))
    return max_depth

def kazmi_syntactic_from_doc(doc):
    result = {}
    for name in feature_names:
        result[name] = f"{0:.5f}"

    #syn_depth and syn_isc are defined per sentence, so calculate per sentence first then take the mean
    sentence_depths = [] #list to store tree depth of each sentence
    sentence_iscs = [] #list to store ISC of each sentence
    for sent in doc.sents:
        if sent.text.strip() == "": #skip whitespace only sentences
           continue
        sentence_depths.append(tree_depth(sent.root))

        isc = 0.0
        for token in sent:
            if token.pos_ == "SCONJ" or token.pos_ == "VERB" or token.pos_ == "AUX":
                isc += 1.0
            if token.dep_ == "acl:relcl" or token.dep_ == "advcl" or token.dep_ == "ccomp" or token.dep_ == "xcomp":
                isc += 1.0
        sentence_iscs.append(isc)

    #paragraph value = unweighted mean of sentence values
    if sentence_depths:
        result["syn_depth"] = f"{(sum(sentence_depths) / len(sentence_depths)):.5f}"
        result["syn_isc"] = f"{(sum(sentence_iscs) / len(sentence_iscs)):.5f}"

    n_arg = 0
    n_adjunct = 0
    for token in doc:
        if token.dep_ in ARG_DEPS:
            n_arg += 1
        if token.dep_ in ADJUNCT_DEPS:
            n_adjunct += 1
    if n_adjunct != 0:
        result["arg_adjunct_ratio"] = f"{(n_arg / n_adjunct):.5f}"

    n_non_space = 0
    dep_counts = {}
    for dep in DEP_REL_TYPES:
        dep_counts[dep] = 0
    for token in doc:
        if token.pos_ == "SPACE":
            continue
        n_non_space += 1
        if token.dep_ in dep_counts:
            dep_counts[token.dep_] += 1
    if n_non_space == 0:
        n_non_space = 1
    for dep in DEP_REL_TYPES:
        name = "dep_" + dep.replace(":", "_")
        result[name] = f"{(dep_counts[dep] / n_non_space):.5f}"

    n_non_punct = 0
    big_counts = {}
    for key in DEP_BIGRAM_KEYS:
        big_counts[key] = 0
    for token in doc:
        if token.pos_ == "SPACE" or token.pos_ == "PUNCT":
            continue
        n_non_punct += 1
        if token.dep_ == "ROOT":
            continue
        head = token.head
        if head.pos_ == "SPACE" or head.pos_ == "PUNCT":
            continue
        if token.i < head.i:
            order = "before"
        else:
            order = "after"
        key = head.pos_ + "_" + token.pos_ + "_" + order
        if key in big_counts:
            big_counts[key] += 1
    if n_non_punct == 0:
        n_non_punct = 1
    for key in DEP_BIGRAM_KEYS:
        result["depbig_" + key] = f"{(big_counts[key] / n_non_punct):.5f}"

    return result

def calculate_kazmi_syntactic(text):
    doc = nlp(text or "")
    return kazmi_syntactic_from_doc(doc)
```

```python
from pathlib import Path
import csv

LEVEL = "paragraph"  # "paragraph" or "sentence" - which data/<LEVEL>/raw-csv to read and data/<LEVEL>/features/ to write
raw_csv_path = Path(f"../../data/{LEVEL}/raw-csv")
out_csv_path = Path(f"../../data/{LEVEL}/features/kazmi-syntactic-csv")
out_csv_path.mkdir(parents=True, exist_ok=True)

for csv_file in sorted(raw_csv_path.glob("*.csv")):
    out_csv_file = out_csv_path / csv_file.name
    if out_csv_file.exists() and out_csv_file.stat().st_size > 0:
       continue
    print(f"Working on file: {csv_file.name}")
    tmp_csv_file = out_csv_file.with_suffix(".tmp")
    with csv_file.open("r",encoding="utf-8") as cf:
       csv_reader = csv.DictReader(cf)
       column_names = list(csv_reader.fieldnames)
       column_names = column_names[:-1] + feature_names + [column_names[-1]]
       rows = list(csv_reader)
    texts = [row["text"] or "" for row in rows]
    with tmp_csv_file.open("w",encoding="utf-8") as of:
       csv_writer = csv.DictWriter(of,fieldnames=column_names)
       csv_writer.writeheader()
       # nlp.pipe parses texts in batches - much faster than calling nlp(text) for every row
       for row, doc in zip(rows, nlp.pipe(texts, batch_size=64)):
          feats = kazmi_syntactic_from_doc(doc)
          for name in feature_names:
             row[name] = feats[name]
          csv_writer.writerow(row)
    tmp_csv_file.replace(out_csv_file)
    #break - During demo we can uncomment and providea demo path
```

```python
# TEST KAZMI SYNTACTIC CODE

text = "She quickly ate sweet apples."
feats = calculate_kazmi_syntactic(text)
print("syn_depth", feats["syn_depth"])
print("syn_isc", feats["syn_isc"])
print("arg_adjunct_ratio", feats["arg_adjunct_ratio"])
print("dep_nsubj", feats["dep_nsubj"])
print("dep_amod", feats["dep_amod"])
print("depbig_VERB_ADV_before", feats["depbig_VERB_ADV_before"])
print("depbig_NOUN_ADJ_before", feats["depbig_NOUN_ADJ_before"])
print("n_features", len(feats))
assert "syn_add" not in feats

# per sentence check - paragraph value must be the mean of the sentence values
text = "Timbaland was one of the most successful producers of the 2000s. His biggest hits include “Promiscuous” by Nelly Furtado, “Work It” by Missy Elliott, and “Cry Me a River” by Justin Timberlake."
doc = nlp(text)
feats = kazmi_syntactic_from_doc(doc)
sentence_feats = [kazmi_syntactic_from_doc(sent.as_doc()) for sent in doc.sents]
print("syn_depth", feats["syn_depth"], [f["syn_depth"] for f in sentence_feats])
print("syn_isc", feats["syn_isc"], [f["syn_isc"] for f in sentence_feats])
for name in ["syn_depth", "syn_isc"]:
    values = [float(f[name]) for f in sentence_feats]
    assert abs(float(feats[name]) - sum(values) / len(values)) < 1e-4

# edge cases - empty text gives 0, not a division by zero
assert calculate_kazmi_syntactic("")["syn_depth"] == "0.00000"
assert calculate_kazmi_syntactic("")["syn_isc"] == "0.00000"
print("all syntactic checks passed")
```
