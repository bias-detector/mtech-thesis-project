Score mean surprisal (bits) for raw-csv texts with ngram -ppl.
Paragraph is split into sentences (spaCy, same as MDL), each sentence is one line for SRILM, sentence means are averaged.
Demo: one paragraph for sir. Bulk: many files per ngram call (20 calls per group).

```python
import csv
import math
import re
import subprocess
import tempfile
from pathlib import Path
import spacy

SRILM_DIR = Path("../../models/srilm").resolve()
LEVEL = "paragraph"  # "paragraph" or "sentence" - which data/<LEVEL>/raw-csv to read and data/<LEVEL>/features/ to write
raw_csv_path = Path(f"../../data/{LEVEL}/raw-csv").resolve() # resolve() - demo cells %cd into SRILM_DIR, relative paths would break after that
# old values (whole paragraph scored as one line, pooled) are in archive-pooled-sm-2026-11-11/mean-surprisal-srilm-csv
out_csv_path = Path(f"../../data/{LEVEL}/features/mean-surprisal-srilm-csv").resolve()
out_csv_path.mkdir(parents=True, exist_ok=True)
(SRILM_DIR / "tmp").mkdir(parents=True, exist_ok=True)

CORPORA = ("brown", "wikitext2", "ptb_wsj", "simplewiki")
ORDERS = (1, 2, 3, 4, 5)

LMS = []
for corpus in CORPORA:
    for order in ORDERS:
        col = f"mean_surprisal_{corpus}_{order}"
        lm = SRILM_DIR / "trained" / f"{corpus}.{order}gram.lm"
        LMS.append((col, lm, order))

WORD_RE = re.compile(r"[A-Za-z0-9]+(?:'[A-Za-z]+)?|[.!?]")

# en_core_web_trf is used only to split paragraph into sentences - same sentence boundaries as mean-mdl.ipynb and mean-surprisal-gpt2.ipynb
# sentences come from the parser, so tagger, attribute_ruler, ner and lemmatizer are disabled
spacy.prefer_gpu() # uses the Mac GPU (MPS) if available
nlp = spacy.load("en_core_web_trf", disable=["tagger", "attribute_ruler", "ner", "lemmatizer"])
print(f"Language model: {nlp.meta['lang']}_{nlp.meta['name']}")
print(f"Version: {nlp.meta['version']}")
print(f"Pipe names: {nlp.pipe_names}")
```

```python
def tokenize_words(text):
    return [m.group(0).lower() for m in WORD_RE.finditer(text)]


def get_sentences(doc):
    #paragraph is split into sentences using the parser (same as mean-mdl.ipynb)
    return list(doc.sents)


def paragraph_to_lines(doc):
    #one line per sentence, so SRILM scores every sentence with its own <s> ... </s>
    #training corpora are also one sentence per line, so the model sees text in the same shape it was trained on
    lines = []
    for sent in get_sentences(doc):
        tokens = tokenize_words(sent.text)
        if tokens: #sentence with no words (only symbols / emoji) is skipped
           lines.append(" ".join(tokens))
    return lines
```

```python
def _parse_sentence_logprobs(ppl_stdout):
    #ngram -debug 1 prints 2 summary lines for every input line:
    #"1 sentences, N words, M OOVs" and "Z zeroprobs, logprob= X ppl= ..."
    results = []
    counts = None
    for raw in ppl_stdout.splitlines():
        line = raw.strip()
        if not line or line.startswith("reading ") or line.startswith("warning:"):
            continue
        if re.match(r"file\s+\S+:\s+\d+\s+sentences?", line): #summary of whole file, all lines are done
            break
        m_w = re.search(r"(\d+)\s+sentences?,\s+(\d+)\s+words?,\s+(\d+)\s+OOVs?", line)
        if m_w:
            counts = (int(m_w.group(1)), int(m_w.group(2)), int(m_w.group(3)))
            continue
        m_lp = re.search(r"(\d+)\s+zeroprobs?,\s+logprob=\s*([-+0-9.eE]+)", line)
        if m_lp and counts is not None:
            n_sents, n_words, n_oovs = counts
            #SRILM scores every word + one </s> per sentence (OOVs / zeroprobs are skipped)
            n_tokens = n_words - n_oovs - int(m_lp.group(1)) + n_sents
            results.append((n_tokens, float(m_lp.group(2))))
            counts = None
    return results


def score_lines_batch(lines, lm_path, order):
    #one ngram -ppl call for many sentence lines, returns (n_tokens, logprob base-10) for every line
    if not lines:
        return []
    lm_rel = Path(lm_path).resolve().relative_to(SRILM_DIR)
    with tempfile.TemporaryDirectory(prefix="srilm_ppl_", dir=SRILM_DIR / "tmp") as tmp:
        text_path = Path(tmp) / "sents.txt"
        text_path.write_text("\n".join(lines) + "\n", encoding="utf-8")
        text_rel = text_path.resolve().relative_to(SRILM_DIR)
        proc = subprocess.run(
            ["ngram", "-lm", str(lm_rel), "-ppl", str(text_rel), "-order", str(order), "-debug", "1", "-unk"],
            cwd=str(SRILM_DIR),
            capture_output=True,
            text=True,
            check=False,
        )
        out = (proc.stdout or "") + "\n" + (proc.stderr or "")

    parsed = _parse_sentence_logprobs(out)
    if len(parsed) != len(lines): #every line must have its own result, else sentences would be matched to wrong values
        raise RuntimeError(f"Expected {len(lines)} ppl lines, got {len(parsed)} (code={proc.returncode}):\n{out[:1500]}")
    return parsed
```

```python
def sentence_surprisal(n_tokens, logprob10):
    #mean surprisal of one sentence in bits = total bits / number of scored tokens
    if n_tokens <= 0: #nothing scored, so no surprisal for this sentence
       return None
    #convert base-10 log prob to bits: log2(p) = log10(p) / log10(2)
    return (-logprob10 / math.log10(2.0)) / n_tokens
```

```python
def calculate_mean_surprisal_batch(paragraph_lines, lm_path, order):
    #paragraph_lines = list of paragraphs, each paragraph = list of its sentence lines
    #all sentences of all paragraphs go to SRILM in one call, then values are grouped back per paragraph
    all_lines = [line for lines in paragraph_lines for line in lines]
    scored = score_lines_batch(all_lines, lm_path, order)

    results = []
    index = 0
    for lines in paragraph_lines:
        sentence_surprisals = [] #list to store mean surprisal of each sentence
        for _ in lines:
            surprisal = sentence_surprisal(*scored[index])
            index += 1
            if surprisal is not None: #skip sentences with nothing scored
               sentence_surprisals.append(surprisal)
        #paragraph surprisal = unweighted mean of sentence surprisals
        if sentence_surprisals:
            mean_surprisal = sum(sentence_surprisals) / len(sentence_surprisals)
        else:
            mean_surprisal = 0
        results.append(f"{mean_surprisal:.5f}")
    return results


def calculate_mean_surprisal(text, lm_path, order):
    return calculate_mean_surprisal_batch([paragraph_to_lines(nlp(text))], lm_path, order)[0]
```

```python
# Demo for sir (step 1): split paragraph into sentences, one line per sentence, run SRILM ngram -ppl
%cd {SRILM_DIR}
demo = "Timbaland was one of the most successful producers of the 2000s. His biggest hits include “Promiscuous” by Nelly Furtado, “Work It” by Missy Elliott, and “Cry Me a River” by Justin Timberlake."
demo_lines = paragraph_to_lines(nlp(demo))
print(f"raw: {demo!r}")
for number, line in enumerate(demo_lines, start=1):
    print(f"sentence {number} tokens ({len(line.split())}): {line}")
(SRILM_DIR / "tmp").mkdir(parents=True, exist_ok=True)
(SRILM_DIR / "tmp" / "demo_sent.txt").write_text("\n".join(demo_lines) + "\n", encoding="utf-8")

!ngram -lm trained/brown.2gram.lm -ppl tmp/demo_sent.txt -order 2 -debug 1 -unk
```

```python
# Demo for sir (step 2): per sentence mean surprisal from the ngram numbers above, then mean over sentences
# for every sentence SRILM prints "1 sentences, N words, M OOVs" and "Z zeroprobs, logprob= X ..."
demo_scored = score_lines_batch(demo_lines, SRILM_DIR / "trained" / "brown.2gram.lm", 2)

demo_sentence_surprisals = [] #list to store mean surprisal of each sentence
for number, (n_tokens, logprob10) in enumerate(demo_scored, start=1):
    # SRILM scores every word + one </s> per sentence (OOVs / zeroprobs are skipped)
    # convert base-10 log prob to bits: log2(p) = log10(p) / log10(2)
    total_bits = -logprob10 / math.log10(2)
    value = sentence_surprisal(n_tokens, logprob10)
    demo_sentence_surprisals.append(value)
    print(f"sentence {number}: logprob10 = {logprob10}, n_tokens = {n_tokens}, total_bits = {total_bits:.5f}, mean_surprisal = {total_bits:.5f} / {n_tokens} = {value:.5f}")

#paragraph surprisal = unweighted mean of sentence surprisals
paragraph_surprisal = sum(demo_sentence_surprisals) / len(demo_sentence_surprisals)
print(f"\nparagraph mean_surprisal = mean of {[round(v, 5) for v in demo_sentence_surprisals]} = {paragraph_surprisal:.5f}")
print(f"calculate_mean_surprisal gives: {calculate_mean_surprisal(demo, SRILM_DIR / 'trained' / 'brown.2gram.lm', 2)}") #same value as the real function

# old method for comparison: whole paragraph as one line (one <s> ... </s>), all tokens pooled
old_n_tokens, old_logprob10 = score_lines_batch([" ".join(demo_lines)], SRILM_DIR / "trained" / "brown.2gram.lm", 2)[0]
print(f"old method (whole paragraph as one line): {sentence_surprisal(old_n_tokens, old_logprob10):.5f}")
```

```python
import gc
import torch

metric_cols = [col for col, _, _ in LMS]
GROUP_ROWS = 20000 # many files go to SRILM together - every ngram call starts a docker container and loads the model, so fewer calls = faster

def process_group(group):
    loaded = [] #(csv file, column names, rows) of every file in this group
    for csv_file in group:
        with csv_file.open("r", encoding="utf-8") as cf:
            reader = csv.DictReader(cf)
            fieldnames = list(reader.fieldnames)
            rows = list(reader)
        loaded.append((csv_file, fieldnames[:-1] + metric_cols + fieldnames[-1:], rows))

    unique_texts = list(dict.fromkeys(row["text"] for _, _, rows in loaded for row in rows))
    # nlp.pipe parses texts in batches - doc is turned into sentence lines right away, so docs are not kept in memory
    lines_by_text = {text: paragraph_to_lines(doc) for text, doc in zip(unique_texts, nlp.pipe(unique_texts, batch_size=64))}
    paragraph_lines = [lines_by_text[text] for text in unique_texts]

    score_maps = {}
    for col, lm_path, order in LMS:
        scores = calculate_mean_surprisal_batch(paragraph_lines, lm_path, order)
        score_maps[col] = dict(zip(unique_texts, scores))

    for csv_file, column_names, rows in loaded:
        out_csv_file = out_csv_path / csv_file.name
        # write to a temp file and rename at the end, so a half written file is never skipped on restart
        tmp_csv_file = out_csv_file.with_suffix(".csv.tmp")
        with tmp_csv_file.open("w", encoding="utf-8", newline="") as of:
            writer = csv.DictWriter(of, fieldnames=column_names)
            writer.writeheader()
            for row in rows:
                for col in metric_cols:
                    row[col] = score_maps[col][row["text"]]
                writer.writerow(row)
        tmp_csv_file.rename(out_csv_file)
    gc.collect(); torch.mps.empty_cache() # free GPU memory used by spaCy before next group

group, group_rows = [], 0
todo = [f for f in sorted(raw_csv_path.glob("*.csv")) if not ((out_csv_path / f.name).exists() and (out_csv_path / f.name).stat().st_size > 0)] # Skip if already exist
print(f"{len(todo)} files to do")
for number, csv_file in enumerate(todo, start=1):
    with csv_file.open("r", encoding="utf-8") as cf:
        group_rows += sum(1 for _ in cf) - 1
    group.append(csv_file)
    if group_rows >= GROUP_ROWS or number == len(todo):
        process_group(group)
        print(f"Done {number}/{len(todo)} files")
        group, group_rows = [], 0
```
