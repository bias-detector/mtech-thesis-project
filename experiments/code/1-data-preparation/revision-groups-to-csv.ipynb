```python
from pathlib import Path

revision_groups_file = Path("../../data/input/revision_groups.jsonl.gz")  # one article per line
raw_csv_path = Path("../../data/paragraph/raw-csv")

raw_csv_path.mkdir(parents=False, exist_ok=True)
```

```python
import gzip, json, csv


with gzip.open(revision_groups_file, "rt", encoding="utf-8") as gz:
    for line in gz:
        article = json.loads(line)
        article_title = article['article_title']
        revision_groups = article['groups']
        csv_file = raw_csv_path / f"{article_title}.csv".replace("/","_")
        with csv_file.open("w",encoding="utf-8") as cf:
          csv_writer = csv.writer(cf)
          csv_writer.writerow(["revision_group_no","revision_depth","text","diff_url"])
          revision_group_no = 1
          for revision_group in revision_groups:
             versions = revision_group["versions"]
             diff_urls = revision_group["diff_urls"]
             for version_num in range(0,len(versions)):
                csv_writer.writerow([revision_group_no,version_num,versions[version_num],diff_urls[version_num]])   
             revision_group_no += 1
#break - To demo 1 file only converted                        
```
